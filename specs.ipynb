```python
import os
import psutil

# Get CPU count
cores = os.cpu_count()
# Get RAM info
ram = psutil.virtual_memory().total / (1024**3)
# Get Disk info
disk = psutil.disk_usage('/').total / (1024**3)

print(f"Allocated CPU Cores: {cores}")
print(f"Allocated RAM: {ram:.2f} GB")
print(f"Allocated Disk Space: {disk:.2f} GB")
```

```python
try:
    with open('/sys/fs/cgroup/cpu.max', 'r') as f:
        quota, period = f.read().strip().split()
        if quota == 'max':
            print("Real CPU Allocation: Unlimited (Using all 32 host cores)")
        else:
            print(f"Real CPU Allocation: {int(quota) / int(period)} Cores")
except Exception:
    # Fallback for older Proxmox cgroup v1 setups
    import os
    q = int(open('/sys/fs/cgroup/cpu/cpu.cfs_quota_us').read().strip())
    p = int(open('/sys/fs/cgroup/cpu/cpu.cfs_period_us').read().strip())
    print(f"Real CPU Allocation: {q / p if q != -1 else 'Unlimited'} Cores")
```
